# Cardio Disease Prediction - Exploratory Data Analysis (EDA)

This notebook explores the dataset to understand the distribution of features and their relationship with the target variable.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load dataset
try:
    df = pd.read_csv('../cardio_train.csv', sep=';')
except FileNotFoundError:
    df = pd.read_csv('cardio_train.csv', sep=';')

df.head()
```

```python
# Basic Stats
df.describe()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Age Distribution (Converted to Years for visualization)
df['age_years'] = (df['age'] / 365).round(1)

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='age_years', hue='cardio', kde=True, bins=30)
plt.title('Age Distribution by Cardio Class')
plt.show()
```

```python
# Correlation Matrix
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.show()
```
